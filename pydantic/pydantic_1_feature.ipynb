# Basics of Pydantic

```python
from pydantic import BaseModel

class Post(BaseModel):
    id: int
    title: str
    category: str
```

```python
post = Post(id=123, title="Hello World", category="Tech")
```

```python
print(post.id)
```

```python
post_dic = {"id": 123, "title": "hello world", "category": "tech"}

post = Post(**post_dic)
```

```python
print(post.title)
```

```python
post = Post(id="123", title="Hello world", category="tech")
```

```python
print(post.id)
```

```python
print(type(post.id))
```

```python
post = Post(id="123a", title="helloworld", category="tech")

print(post.id)
```

```python
post = Post(id="123", title="Hello world", category=1234)
```
